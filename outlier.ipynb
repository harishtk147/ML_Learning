```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

data = np.array([250,275,300,320,325,330,350,360,1000])

z_scores = np.abs(stats.zscore(data))

threshold = 3

outlier = np.where(z_scores > threshold)[0]
print("Outliers (Indices):", outlier)

colors = "red"

plt.figure(figsize=(8,6))
plt.scatter(np.arange(len(data)),data, c=colors)
plt.show()
```
