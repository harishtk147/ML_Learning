```python
import pandas as pd
```

```python
df=pd.read_csv('demo_dataset.csv')
df
```

```python
df.plot(kind='bar',x='age',y='salary')
```

```python
df.rename(columns={'age':'Age'},inplace=True)
df
```

```python
df.fillna({'Age':df['Age'].mean()},inplace=True)
```

```python
df.corr()
```

```python
df['gender']=['M','F','M','M','F','M','M','F','M','M']
```

```python
g = df.groupby('gender')['Age'].sum()
g
```
