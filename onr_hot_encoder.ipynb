```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder()
```

```python
data = {'Fruit' : ['Apple', 'Banana', 'Orange', 'Grapes', 'Mango'],
        'Color' : ['Red', 'Yellow', 'Orange', 'Green', 'Yellow']
}
```

```python
df = pd.DataFrame(data)
```

```python
encoder_color = encoder.fit_transform(df[['Color']]).toarray()
```

```python
encoder_df = pd.DataFrame(encoder_color, columns=encoder.get_feature_names_out(['Color']))
```

```python
pd.concat([df,encoder_df], axis=1)
```
